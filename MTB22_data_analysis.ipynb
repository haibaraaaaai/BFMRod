```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.mixture import GaussianMixture
from scipy.stats import norm

# Apply desired publication formatting
plt.rcParams.update({
    'font.size': 12,
    'axes.labelsize': 22,
    'axes.titlesize': 22,
    'xtick.labelsize': 20,
    'ytick.labelsize': 20,
    'legend.fontsize': 12
})

BASE_PATH = os.getcwd()

# Original trace set, with comments preserved
target_files = {
    "20250311": [
        "file_stepfinder_outputs",
        "file1_stepfinder_outputs", #anysotropy concerns
        "file2_stepfinder_outputs", #anysotropy concerns
        "file3_stepfinder_outputs" #anysotropy concerns
    ],
    "20250313": [
        "file0224_stepfinder_outputs",
        #"file02225_stepfinder_outputs", #ALWAYS BLOCKED
        "fileMinArab2_stepfinder_outputs",#anysotropy concerns*
        "fileMinArab3_stepfinder_outputs", #anysotropy concerns*
        "fileMinArab4_stepfinder_outputs"
    ], 
    "20250320": [
        "file_stepfinder_outputs",
        "file4_stepfinder_outputs",
        "file5_stepfinder_outputs", #anysotropy concerns
        "file8_stepfinder_outputs",
        "file9_stepfinder_outputs",
        "file10_stepfinder_outputs",
        "file12_stepfinder_outputs",
        "file13_stepfinder_outputs",
        "file15_stepfinder_outputs",
        "file16_stepfinder_outputs",
        "file17_stepfinder_outputs"
    ],
    "20250325": [
        "file1_stepfinder_outputs",
        "file3_stepfinder_outputs",
        "file5_stepfinder_outputs",
        "file7_stepfinder_outputs",
        "file9_stepfinder_outputs",
        "file10_stepfinder_outputs",
        "file11_stepfinder_outputs",
        "file12_stepfinder_outputs",
        "file13_stepfinder_outputs"
    ]
}

# Process and plot per trace
for date_str, folder_list in target_files.items():
    for folder_name in folder_list:
        base_folder = os.path.join(BASE_PATH, date_str, "results", folder_name)
        smooth_path = os.path.join(base_folder, "smoothed_angular_speed_trace.csv")
        stepfit_path = os.path.join(base_folder, "itot_speed_trace_stepfit.csv")
        
        if not os.path.exists(smooth_path) or not os.path.exists(stepfit_path):
            print(f"⚠️ Missing one or both files in {base_folder}")
            continue

        try:
            # Load smoothed trace
            smoothed_df = pd.read_csv(smooth_path, usecols=["Smoothed Speed (Hz)"])
            smoothed_speeds = pd.to_numeric(smoothed_df["Smoothed Speed (Hz)"], errors='coerce').dropna()

            # Load step-fitted trace
            stepfit_df = pd.read_csv(stepfit_path, usecols=["Step-Fitted Speed (Hz)"])
            stepfit_speeds = pd.to_numeric(stepfit_df["Step-Fitted Speed (Hz)"], errors='coerce').dropna()

            # Use shared bin range
            combined = pd.concat([smoothed_speeds, stepfit_speeds])
            bin_min, bin_max = combined.min(), combined.max()
            bins = np.linspace(bin_min, bin_max, 40)

            # Plot
            fig, axs = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

            axs[0].hist(smoothed_speeds, bins=bins, color='skyblue', alpha=0.7)
            axs[0].set_title("Smoothed Speed")
            axs[0].set_xlabel("Speed (Hz)")
            axs[0].set_ylabel("Count")
            axs[0].grid(True)

            axs[1].hist(stepfit_speeds, bins=bins, color='salmon', alpha=0.7)
            axs[1].set_title("Step-Fitted Speed")
            axs[1].set_xlabel("Speed (Hz)")
            axs[1].grid(True)

            fig.suptitle(f"{date_str} — {folder_name}", fontsize=24)
            plt.tight_layout(rect=[0, 0, 1, 0.95])

            # Save → Display → Close
            output_path = os.path.join(base_folder, f"{folder_name}_histograms.png")
            fig.savefig(output_path, dpi=300)
            display(fig)
            plt.close(fig)
            print(f"✅ Saved and displayed: {output_path}")

        except Exception as e:
            print(f"❌ Error processing {folder_name}: {e}")
```

```python
if len(all_stepfit) == 0:
    print("⚠️ No step-fitted speed data to analyze.")
else:
    X = np.array(all_stepfit).reshape(-1, 1)
    models = {}
    print("📊 GMM AIC/BIC analysis:\n")

    for n_components in [1, 2, 3]:
        gmm = GaussianMixture(n_components=n_components, random_state=42)
        gmm.fit(X)
        aic = gmm.aic(X)
        bic = gmm.bic(X)
        models[n_components] = {"model": gmm, "AIC": aic, "BIC": bic}
        print(f"{n_components} components: AIC = {aic:.2f}, BIC = {bic:.2f}")

    # Determine best by AIC and BIC
    best_aic = min(models.items(), key=lambda x: x[1]["AIC"])[0]
    best_bic = min(models.items(), key=lambda x: x[1]["BIC"])[0]

    print(f"\n✅ Best by AIC: {best_aic} components")
    print(f"✅ Best by BIC: {best_bic} components (more conservative)")
```

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 10), sharex=False, sharey=True)
X_smooth = np.array(all_smoothed).reshape(-1, 1)
X_step = np.array(all_stepfit).reshape(-1, 1)

bins_smooth = np.linspace(X_smooth.min(), X_smooth.max(), 40)
bins_step = np.linspace(X_step.min(), X_step.max(), 40)

for row, n_components in enumerate([2, 3]):
    for col, (X, label, bins) in enumerate([
        (X_smooth, "Smoothed Speed (Hz)", bins_smooth),
        (X_step, "Step-Fitted Speed (Hz)", bins_step)
    ]):
        gmm = GaussianMixture(n_components=n_components, random_state=42)
        gmm.fit(X)

        weights = gmm.weights_
        means = gmm.means_.flatten()
        stds = np.sqrt(gmm.covariances_).flatten()
        order = np.argsort(means)
        weights, means, stds = weights[order], means[order], stds[order]

        # Print stats
        print(f"\n📊 {label} — {n_components} Components:")
        for i in range(n_components):
            print(f"  Gaussian {i+1}: mean = {means[i]:.2f} Hz, std = {stds[i]:.2f} Hz, weight = {weights[i]:.3f}")

        # Plot histogram
        ax = axs[row, col]
        ax.hist(X.flatten(), bins=bins, density=True, alpha=0.4, color='gray', label='Histogram')

        # Plot GMM components
        x_vals = np.linspace(bins.min(), bins.max(), 1000)
        for i in range(n_components):
            y = weights[i] * norm.pdf(x_vals, means[i], stds[i])
            ax.plot(x_vals, y, lw=2, label=f"Gaussian {i+1}")
            ax.axvline(means[i], color='k', linestyle='--', lw=1)

        ax.set_title(f"{label} — GMM ({n_components} comp)")
        ax.set_xlabel("Speed (Hz)")
        if col == 0:
            ax.set_ylabel("Density")
        ax.grid(True)
        ax.legend()

plt.suptitle("Global Speed Distributions — GMM Fits (Smoothed vs. Step-Fitted)", fontsize=24)
plt.tight_layout(rect=[0, 0, 1, 0.96])
display(fig)
plt.close(fig)
```

```python
# Helper to fit GMM and return sorted means/stds/weights
def fit_gmm(data_array, n_components=2):
    data_array = np.array(data_array).reshape(-1, 1)
    gmm = GaussianMixture(n_components=n_components, random_state=42)
    gmm.fit(data_array)
    means = gmm.means_.flatten()
    stds = np.sqrt(gmm.covariances_).flatten()
    weights = gmm.weights_.flatten()
    order = np.argsort(means)
    return means[order], stds[order], weights[order]

# Loop over all target files
for date_str, folder_list in target_files.items():
    for folder_name in folder_list:
        folder_path = os.path.join(BASE_PATH, date_str, "results", folder_name)
        smoothed_path = os.path.join(folder_path, "smoothed_angular_speed_trace.csv")
        stepfit_path = os.path.join(folder_path, "stepfit_dwell_time_summary.csv")

        if not os.path.exists(smoothed_path) or not os.path.exists(stepfit_path):
            print(f"⚠️ Missing file(s) for {folder_name}")
            continue

        try:
            # Load smoothed and stepfitted data
            smoothed_data = pd.read_csv(smoothed_path, usecols=["Smoothed Speed (Hz)"])
            smoothed = smoothed_data["Smoothed Speed (Hz)"].dropna().values

            stepfit_data = pd.read_csv(stepfit_path, usecols=["Rounded Level (Hz)"])
            stepfit = stepfit_data["Rounded Level (Hz)"].dropna().values

            # Fit 2-GMMs
            sm_means, sm_stds, sm_weights = fit_gmm(smoothed, 2)
            st_means, st_stds, st_weights = fit_gmm(stepfit, 2)

            # SHOW GMM fits (not saved)
            fig, axs = plt.subplots(1, 2, figsize=(14, 5))
            for ax, data, means, stds, weights, label in zip(
                axs,
                [smoothed, stepfit],
                [sm_means, st_means],
                [sm_stds, st_stds],
                [sm_weights, st_weights],
                ["Smoothed", "Step-Fitted"]
            ):
                bins = np.linspace(min(data), max(data), 40)
                ax.hist(data, bins=bins, density=True, alpha=0.5, color='gray', label='Histogram')
                x_vals = np.linspace(min(data), max(data), 1000)
                for i in range(2):
                    ax.plot(x_vals, weights[i] * norm.pdf(x_vals, means[i], stds[i]), lw=2, label=f"Gaussian {i+1}")
                    ax.axvline(means[i], color='k', linestyle='--', lw=1)
                ax.set_title(f"{label} — GMM (2 comp)")
                ax.set_xlabel("Speed (Hz)")
                ax.grid(True)
                ax.legend()
            plt.suptitle(f"{folder_name} — Smoothed vs Step-Fitted GMM")
            plt.tight_layout(rect=[0, 0, 1, 0.95])
            display(fig)
            plt.close(fig)

            # NORMALIZATION
            state1_index = np.argmin(np.abs(sm_means - main_peak))
            norm_factor = sm_means[state1_index]

            norm_smoothed = smoothed / norm_factor
            norm_stepfit = stepfit / norm_factor

            # SAVE normalized values
            pd.DataFrame(norm_smoothed, columns=["Normalized Smoothed Speed"]).to_csv(
                os.path.join(folder_path, "normalized_smoothed_speeds.csv"), index=False
            )
            pd.DataFrame(norm_stepfit, columns=["Normalized Step-Fitted Speed"]).to_csv(
                os.path.join(folder_path, "normalized_stepfit_speeds.csv"), index=False
            )

            # PLOT & SAVE histograms

            # Smoothed histogram
            fig1, ax1 = plt.subplots(figsize=(10, 4))
            ax1.hist(norm_smoothed, bins=40, density=True, alpha=0.6, color='orangered')
            ax1.set_title(f"Normalized Smoothed Speeds — {folder_name}")
            ax1.set_xlabel("Normalized Speed")
            ax1.set_ylabel("Density")
            ax1.grid(True)
            plt.tight_layout()
            display(fig1)
            fig1.savefig(os.path.join(folder_path, "normalized_smoothed_hist.png"))
            plt.close(fig1)

            # Step-fitted histogram
            fig2, ax2 = plt.subplots(figsize=(10, 4))
            ax2.hist(norm_stepfit, bins=40, density=True, alpha=0.6, color='steelblue')
            ax2.set_title(f"Normalized Step-Fitted Speeds — {folder_name}")
            ax2.set_xlabel("Normalized Speed")
            ax2.set_ylabel("Density")
            ax2.grid(True)
            plt.tight_layout()
            display(fig2)
            fig2.savefig(os.path.join(folder_path, "normalized_stepfit_hist.png"))
            plt.close(fig2)

        except Exception as e:
            print(f"❌ Error processing {folder_name}: {e}")
```

```python

# === Pooled output across all days ===
pooled_norm_smoothed = []
pooled_norm_stepfit = []

for date_str, folder_list in target_files.items():
    smoothed_all = []

    # Step 1: Pool smoothed values for the day
    for folder_name in folder_list:
        folder_path = os.path.join(BASE_PATH, date_str, "results", folder_name)
        smoothed_path = os.path.join(folder_path, "smoothed_angular_speed_trace.csv")

        if os.path.exists(smoothed_path):
            try:
                smoothed_data = pd.read_csv(smoothed_path, usecols=["Smoothed Speed (Hz)"])
                smoothed_all.extend(smoothed_data["Smoothed Speed (Hz)"].dropna().tolist())
            except Exception as e:
                print(f"⚠️ Could not read smoothed in {folder_name}: {e}")

    if len(smoothed_all) < 10:
        print(f"⚠️ Skipping {date_str}: insufficient pooled smoothed data")
        continue

    # Step 2: Fit GMM to all smoothed speeds of the day
    means, _, _ = fit_gmm(smoothed_all, n_components=2)
    state1_idx = np.argmin(np.abs(means - main_peak))
    norm_factor = means[state1_idx]
    print(f"📅 {date_str} — Normalizing by state 1 peak: {norm_factor:.2f} Hz")

    # Step 3: Normalize all traces of that day
    for folder_name in folder_list:
        folder_path = os.path.join(BASE_PATH, date_str, "results", folder_name)
        smoothed_path = os.path.join(folder_path, "smoothed_angular_speed_trace.csv")
        stepfit_path = os.path.join(folder_path, "stepfit_dwell_time_summary.csv")

        try:
            # Smoothed
            if os.path.exists(smoothed_path):
                smoothed_data = pd.read_csv(smoothed_path, usecols=["Smoothed Speed (Hz)"])
                norm_smoothed = smoothed_data["Smoothed Speed (Hz)"].dropna() / norm_factor
                pooled_norm_smoothed.extend(norm_smoothed.tolist())

            # Step-fitted
            if os.path.exists(stepfit_path):
                stepfit_data = pd.read_csv(stepfit_path, usecols=["Rounded Level (Hz)"])
                norm_stepfit = stepfit_data["Rounded Level (Hz)"].dropna() / norm_factor
                pooled_norm_stepfit.extend(norm_stepfit.tolist())

        except Exception as e:
            print(f"❌ Error processing {folder_name}: {e}")
```
